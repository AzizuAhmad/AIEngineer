# HANDLING MISSING VALUES

```python
dfPyspark = spark.read.csv("test2.csv", header= True,inferSchema=True)
dfPyspark.show()
```

```python
# drop columns
dfPyspark.drop('Name').show()
```

```python
dfPyspark.show()
```

```python
#  drop all null value
dfPyspark.na.drop().show()
```

```python
# any == how

dfPyspark.na.drop(how="all").show()
```

```python
# threshold
dfPyspark.na.drop(how="all",thresh=2).show()
```

```python
# subset
dfPyspark.na.drop(how="any",subset=['Age']).show()
```

```python
# Filling missing value
dfPyspark.na.fill("Missing Values", ["Name"]).show()
```

```python
dfPyspark.na.fill(30, ["Age"]).show()
```

```python
dfPyspark.show()
```

```python
from pyspark.ml.feature import Imputer

imputer = Imputer(
    inputCols=["age", "Experience","Salary"],
    outputCols=["{}_imputed".format(c) for c in ["age", "Experience","Salary"]]
).setStrategy("mean")
```

```python
imputer.fit(dfPyspark).transform(dfPyspark).show()
```

# Example of pyspark ML

```python
import pyspark
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("ML").getOrCreate()
```

```python
# read the dataset
training = spark.read.csv("test1.csv", header = True, inferSchema= True)
```

```python
training.show()
```

```python
training.printSchema()
```

```python
training.columns
```

```python
from pyspark.ml.feature import VectorAssembler
featureAssambler = VectorAssembler(inputCols=["age","Experience"],outputCol="Independent Features")
```

```python
output = featureAssambler.transform(training)
output.show()
```

```python
output.columns
```

```python
output.dtypes
```

```python
finaldata = output.select("Independent Features","Salary")
finaldata.show()
```

```python
from pyspark.ml.regression import LinearRegression
# train test split
train_data,test_data = finaldata.randomSplit([0.75,0.25],42)
regressor = LinearRegression(featuresCol="Independent Features", labelCol= "Salary")
regressor = regressor.fit(train_data)
```

```python
# coefisien

regressor.coefficients
```

```python
# intercepts

regressor.intercept
```

```python
predResult = regressor.evaluate(test_data)
predResult.predictions.show()
```

```python
predResult.meanAbsoluteError, predResult.meanSquaredError
```
